```python
# Mini-Task 3: Edit this function to properly call the OpenAI API to get ChatGPT's Response.
def get_insights_from_chatgpt(client, linkedin_post):
  # Mini-Task 4: Get a ChatGPT prompt from a function.
    prompt = get_chatgpt_prompt(linkedin_post)

    # Call the OpenAI API
    try:
      response = client.chat.completions.create(
        model="gpt-5",
        messages=[
          {"role": "system", "content": "You are an expert in social media marketing and content analysis."},
          {"role": "user", "content": prompt}]
      )

      # Extract response text
      return response.choices[0].message.content
    except Exception as e:
      return f"Error: {e}"
```

```python
def main():
    # Mini-Task 4: Get the LinkedIn post from the user.
    linkedin_post = input("Please paste your LinkedIn post text here: ")

    # Mini-Task 5: Get OpenAI API Key here from the user.
    openai_api_key = input("Please enter your OpenAI API key: ")

    # Initialize the OpenAI client
    client = openai.OpenAI(api_key=openai_api_key)

    # Mini-Task 6: Call get_insights_from_chatgpt function with a client and a LinkedIn post
    insights = get_insights_from_chatgpt(client, linkedin_post)

    # Display the insights to the user
    print("\nInsights from ChatGPT:")
    print(insights)
```

```python
main()
```
